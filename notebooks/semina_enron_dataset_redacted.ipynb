```python
import pandas as pd
import re
from pathlib import Path

FREEZE_DIR = Path("/content/drive/MyDrive/forensic_email_project/frozen_release_v1")

# Load data
full = pd.read_csv(FREEZE_DIR / "data" / "Enron_100000_full.csv")

# Load rules (already parsed earlier)
rules_dict = {
    "core": rules["core_indicators"],
    "urgency": rules["urgency_indicators"],
    "action": rules["action_indicators"],
    "benign": rules["benign_context_indicators"]
}

regex_rules = {
    "url": re.compile(rules["regex_indicators"]["url"], re.IGNORECASE),
    "url_shortener": re.compile(rules["regex_indicators"]["url_shorteners"], re.IGNORECASE),
    "email_in_body": re.compile(rules["regex_indicators"]["email_in_body"], re.IGNORECASE)
}

# Helper function
def contains_any(text, phrases):
    text = str(text).lower()
    return any(p in text for p in phrases)

# Build triggers
records = []

for _, row in full.iterrows():
    text = f"{row.get('subject','')} {row.get('body','')}".lower()

    r = {
        "id": row["id"],
        "y_bin": row["y_bin"]
    }

    # Core categories
    r["R_core"] = int(contains_any(text, rules_dict["core"]))
    r["R_urgency"] = int(contains_any(text, rules_dict["urgency"]))
    r["R_action"] = int(contains_any(text, rules_dict["action"]))
    r["R_benign"] = int(contains_any(text, rules_dict["benign"]))

    # Regex
    r["R_url"] = int(bool(regex_rules["url"].search(text)))
    r["R_url_short"] = int(bool(regex_rules["url_shortener"].search(text)))
    r["R_email"] = int(bool(regex_rules["email_in_body"].search(text)))

    # Score (from YAML weights)
    score = (
        r["R_core"] * rules["weights"]["core"] +
        r["R_urgency"] * rules["weights"]["urgency"] +
        r["R_action"] * rules["weights"]["action"] +
        r["R_url"] * rules["weights"]["url"] +
        r["R_url_short"] * rules["weights"]["url_shortener"]
    )

    r["rule_score"] = score
    r["rule_label"] = int(score >= rules["thresholds"]["phish_score_threshold"])

    records.append(r)

rule_df = pd.DataFrame(records)

TRIGGER_PATH = FREEZE_DIR / "results" / "rule_triggers_100k.csv"
rule_df.to_csv(TRIGGER_PATH, index=False)

print("Saved:", TRIGGER_PATH)
rule_df.head()
```

```python
import pandas as pd
import re
from pathlib import Path

QUAL_PATH = FREEZE_DIR / "results" / "qualitative_error_analysis_notes.csv"

qual = pd.read_csv(QUAL_PATH)

def classify_error(row):
    text = f"{row.get('subject','')} {row.get('body_preview','')}".lower()

    newsletter_terms = [
        "newsletter", "digest", "daily digest", "word of the day",
        "amazon.com", "completecrafts", "enerfax", "chapters.ca",
        "unsubscribe", "you received this email", "online report"
    ]

    promo_terms = [
        "gift", "claim your", "dvd", "hot new releases", "priced to sell",
        "click here", "shipping and handling"
    ]

    system_access_terms = [
        "account access", "guest password", "authorized user", "log in",
        "login", "new url", "access", "password"
    ]

    corporate_terms = [
        "credit worthiness", "study", "trader", "eol", "counterparty",
        "reporting production system", "business highlights"
    ]

    political_terms = [
        "recount", "campaign", "bush", "cheney"
    ]

    if any(t in text for t in newsletter_terms):
        return (
            "Routine newsletter/digest or automated external information feed.",
            "newsletter_or_automated_bulk_message",
            "Model likely weighted bulk/external-message structure as suspicious although content appears routine."
        )

    if any(t in text for t in promo_terms):
        return (
            "Promotional or commercial bulk message.",
            "promotional_bulk_message",
            "High-confidence error likely caused by promotional formatting, external sender patterns, or bulk-message structure."
        )

    if any(t in text for t in system_access_terms):
        return (
            "Operational account/access notification.",
            "system_or_account_access_message",
            "Access-related terminology may resemble security-sensitive communication but content appears operational."
        )

    if any(t in text for t in corporate_terms):
        return (
            "Business-operational communication with domain-specific terminology.",
            "domain_specific_business_context",
            "Model likely missed or over-weighted specialized business context not well represented in sparse text features."
        )

    if any(t in text for t in political_terms):
        return (
            "Political or campaign-related bulk communication.",
            "external_political_bulk_message",
            "Weak-labeling or classifier signal may treat external mass communication as suspicious despite non-forensic routine content."
        )

    if row["case_type"] == "false_negative":
        return (
            "Positive weak-label case with limited or context-dependent lexical evidence.",
            "weak_label_or_context_missing",
            "Likely requires thread-level or metadata-level context beyond the available message preview."
        )

    return (
        "High-scoring false positive requiring manual confirmation.",
        "manual_review_required",
        "No automatic category confidently assigned."
    )

filled = qual.copy()

for idx, row in filled.iterrows():
    obs, source, note = classify_error(row)
    filled.at[idx, "manual_observation"] = obs
    filled.at[idx, "likely_error_source"] = source
    filled.at[idx, "reviewer_note"] = note

filled.to_csv(QUAL_PATH, index=False)

print("Updated:", QUAL_PATH)
filled[[
    "case_type", "subject", "manual_observation",
    "likely_error_source", "reviewer_note"
]]
```

```python
import os

def print_tree(start_path, max_depth=3, prefix=""):
    start_path = str(start_path)
    if max_depth < 0:
        return
    try:
        items = sorted(os.listdir(start_path))
    except:
        return

    for i, item in enumerate(items):
        path = os.path.join(start_path, item)
        connector = "└── " if i == len(items)-1 else "├── "
        print(prefix + connector + item)
        if os.path.isdir(path):
            extension = "    " if i == len(items)-1 else "│   "
            print_tree(path, max_depth-1, prefix + extension)

print_tree("/content/drive/MyDrive/forensic_email_project/frozen_release_v1", max_depth=3)
```

```python
#Step 6B — Load tokenizer/model and generate DistilBERT probabilities
import torch
from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, DataCollatorWithPadding, Trainer, TrainingArguments

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

def load_text_splits(scale_name):
    train_df = pd.read_csv(DATA_DIR / f"{scale_name}_full.csv")[["id", "text", "y_bin"]]
    train_ids = pd.read_csv(SPLITS_DIR / scale_name / "train_ids.csv")
    val_ids   = pd.read_csv(SPLITS_DIR / scale_name / "val_ids.csv")
    test_ids  = pd.read_csv(SPLITS_DIR / scale_name / "test_ids.csv")

    train = train_df.merge(train_ids, on="id", how="inner")
    val   = train_df.merge(val_ids, on="id", how="inner")
    test  = train_df.merge(test_ids, on="id", how="inner")
    return train, val, test

def tokenize_batch_for_infer(batch, tokenizer, max_length=256):
    return tokenizer(batch["text"], truncation=True, max_length=max_length)

def get_distilbert_probs(scale_name, split_name, df):
    model_dir = MODELS_DIR / f"distilbert_{scale_name}_weighted_best"

    tokenizer = AutoTokenizer.from_pretrained(model_dir)
    model = AutoModelForSequenceClassification.from_pretrained(model_dir).to(DEVICE)

    ds = Dataset.from_pandas(df[["text", "y_bin"]].rename(columns={"y_bin": "label"}).reset_index(drop=True))
    ds_tok = ds.map(lambda b: tokenize_batch_for_infer(b, tokenizer), batched=True, remove_columns=["text"])

    data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

    args = TrainingArguments(
        output_dir=str(ROOT / "tmp_infer"),
        per_device_eval_batch_size=32,
        report_to="none",
        disable_tqdm=True,
    )

    trainer = Trainer(
        model=model,
        args=args,
        data_collator=data_collator,
    )

    pred_out = trainer.predict(ds_tok)
    logits = pred_out.predictions
    probs = torch.softmax(torch.tensor(logits), dim=1).numpy()

    out = pd.DataFrame({
        "id": df["id"].values,
        "distil_prob_0": probs[:, 0],
        "distil_prob_1": probs[:, 1],
        "distil_logit_0": logits[:, 0],
        "distil_logit_1": logits[:, 1],
        "y_bin": df["y_bin"].values
    })

    return out
```

```python
#Step 5B — Run fused TF-IDF + metadata Logistic Regression
import numpy as np
from scipy.sparse import hstack, csr_matrix
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, average_precision_score, roc_auc_score

def run_text_metadata_fusion_lr(scale_name, max_features=20000):
    train_df, val_df, test_df = build_text_metadata_splits(scale_name)

    # text
    X_train_text = train_df["text"].fillna("")
    X_test_text  = test_df["text"].fillna("")

    # labels
    y_train = train_df["y_bin"].astype(int)
    y_test  = test_df["y_bin"].astype(int)

    # metadata
    X_train_meta = train_df[METADATA_FEATURES]
    X_test_meta  = test_df[METADATA_FEATURES]

    # TF-IDF text branch
    vec = TfidfVectorizer(max_features=max_features, stop_words="english")
    Xtr_text = vec.fit_transform(X_train_text)
    Xte_text = vec.transform(X_test_text)

    # metadata branch
    imputer = SimpleImputer(strategy="median")
    scaler = StandardScaler()

    Xtr_meta = imputer.fit_transform(X_train_meta)
    Xte_meta = imputer.transform(X_test_meta)

    Xtr_meta = scaler.fit_transform(Xtr_meta)
    Xte_meta = scaler.transform(Xte_meta)

    Xtr_meta = csr_matrix(Xtr_meta)
    Xte_meta = csr_matrix(Xte_meta)

    # fuse
    Xtr_fused = hstack([Xtr_text, Xtr_meta])
    Xte_fused = hstack([Xte_text, Xte_meta])

    clf = LogisticRegression(max_iter=2000, class_weight="balanced", random_state=42)
    clf.fit(Xtr_fused, y_train)

    pred = clf.predict(Xte_fused)
    prob = clf.predict_proba(Xte_fused)[:, 1]

    rep = classification_report(y_test, pred, output_dict=True)

    row = {
        "scale": scale_name,
        "model_family": "LogisticRegression",
        "variant": "text_plus_metadata",
        "macro_f1": rep["macro avg"]["f1-score"],
        "class1_recall": rep["1"]["recall"],
        "class1_f1": rep["1"]["f1-score"],
        "pr_auc": average_precision_score(y_test, prob),
        "roc_auc": roc_auc_score(y_test, prob),
    }

    artifacts = {
        "vectorizer": vec,
        "imputer": imputer,
        "scaler": scaler,
        "classifier": clf,
    }

    return row, artifacts
```

```python
#Step 3A — Metadata extraction helpers

import pandas as pd
import numpy as np
import re
from pathlib import Path
from email.utils import parsedate_to_datetime

ROOT = Path("/content/drive/MyDrive/forensic_email_project")
DATA_DIR = ROOT / "data"

def get_header_value(message: str, header: str) -> str:
    if not isinstance(message, str):
        return ""
    m = re.search(rf"(?im)^{header}:\s*(.*)$", message)
    return m.group(1).strip() if m else ""

def extract_addresses(header_value: str):
    if not isinstance(header_value, str):
        return []
    return re.findall(r"[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,}", header_value, flags=re.I)

def count_unique_domains(addresses):
    domains = []
    for a in addresses:
        if "@" in a:
            domains.append(a.split("@")[-1].lower())
    return len(set(domains))

def is_internal_address(addr: str):
    return isinstance(addr, str) and addr.lower().endswith("@enron.com")

def safe_ratio(a, b):
    return float(a) / float(b) if b not in [0, None] else 0.0

def parse_email_datetime(date_str):
    try:
        dt = parsedate_to_datetime(date_str)
        return dt
    except Exception:
        return None

def tokenize_len(text):
    if not isinstance(text, str) or not text.strip():
        return 0
    return len(text.split())

def has_attachment_mention(text):
    if not isinstance(text, str):
        return 0
    text = text.lower()
    patterns = [
        "attached",
        "attachment",
        "attached file",
        "see attached",
        "enclosed"
    ]
    return int(any(p in text for p in patterns))

def extract_metadata_features(df):
    rows = []

    for _, r in df.iterrows():
        message = r.get("message", "")
        subject = r.get("subject", "")
        body = r.get("body", "")
        text = r.get("text", "")

        from_v = get_header_value(message, "From")
        to_v   = get_header_value(message, "To")
        cc_v   = get_header_value(message, "Cc")
        bcc_v  = get_header_value(message, "Bcc")
        date_v = get_header_value(message, "Date")

        from_addrs = extract_addresses(from_v)
        to_addrs   = extract_addresses(to_v)
        cc_addrs   = extract_addresses(cc_v)
        bcc_addrs  = extract_addresses(bcc_v)

        all_recips = to_addrs + cc_addrs + bcc_addrs
        all_parties = from_addrs + all_recips

        sender_internal = int(any(is_internal_address(a) for a in from_addrs))
        sender_external = int(len(from_addrs) > 0 and not all(is_internal_address(a) for a in from_addrs))

        internal_only_thread = int(len(all_parties) > 0 and all(is_internal_address(a) for a in all_parties))
        mixed_domain_thread = int(any(is_internal_address(a) for a in all_parties) and any(not is_internal_address(a) for a in all_parties))

        subj_lower = subject.lower() if isinstance(subject, str) else ""

        dt = parse_email_datetime(date_v)
        if dt is not None:
            hour_of_day = int(dt.hour)
            day_of_week = int(dt.weekday())   # 0=Mon ... 6=Sun
            is_weekend = int(day_of_week >= 5)
            is_business_hours = int(8 <= hour_of_day <= 18 and day_of_week < 5)
        else:
            hour_of_day = -1
            day_of_week = -1
            is_weekend = 0
            is_business_hours = 0

        subject_char_len = len(subject) if isinstance(subject, str) else 0
        body_char_len = len(body) if isinstance(body, str) else 0
        subject_token_len = tokenize_len(subject)
        body_token_len = tokenize_len(body)

        row = {
            "id": r["id"],
            "label3_v3": r["label3_v3"],
            "y_bin": r["y_bin"],

            # header / communication structure
            "is_internal_sender": sender_internal,
            "is_external_sender": sender_external,
            "num_to_recipients": len(to_addrs),
            "num_cc_recipients": len(cc_addrs),
            "num_bcc_recipients": len(bcc_addrs),
            "has_cc": int(len(cc_addrs) > 0),
            "has_bcc": int(len(bcc_addrs) > 0),
            "is_reply_subject": int(subj_lower.startswith("re:")),
            "is_forward_subject": int(subj_lower.startswith("fw:") or subj_lower.startswith("fwd:")),
            "num_recipient_domains": count_unique_domains(all_recips),
            "is_internal_only_thread": internal_only_thread,
            "is_mixed_domain_thread": mixed_domain_thread,

            # message shape
            "subject_char_len": subject_char_len,
            "body_char_len": body_char_len,
            "subject_token_len": subject_token_len,
            "body_token_len": body_token_len,
            "body_to_subject_len_ratio": safe_ratio(body_char_len, max(subject_char_len, 1)),
            "has_empty_subject": int(subject_char_len == 0),
            "has_short_body": int(body_token_len < 20),

            # temporal
            "hour_of_day": hour_of_day,
            "is_business_hours": is_business_hours,
            "day_of_week": day_of_week,
            "is_weekend": is_weekend,

            # conservative flags
            "has_attachment_mention": has_attachment_mention(text),
            "has_multiple_recipients": int(len(all_recips) > 1),
        }

        rows.append(row)

    return pd.DataFrame(rows)
```

```python
# Cell 10 — DistilBERT helpers
from datasets import Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    DataCollatorWithPadding,
    TrainingArguments,
    Trainer
)
import torch.nn as nn
import torch
import numpy as np
from sklearn.metrics import f1_score, recall_score

tokenizer = AutoTokenizer.from_pretrained(CONFIG["distilbert_model"])
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

def tokenize_batch(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        max_length=CONFIG["distilbert_max_length"]
    )

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=1)
    return {
        "macro_f1": f1_score(labels, preds, average="macro"),
        "class1_recall": recall_score(labels, preds, pos_label=1),
        "class1_f1": f1_score(labels, preds, pos_label=1),
    }

class WeightedTrainer(Trainer):
    def __init__(self, class_weights, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.class_weights = class_weights

    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels = inputs["labels"]

        outputs = model(
            input_ids=inputs["input_ids"],
            attention_mask=inputs["attention_mask"]
        )

        logits = outputs.logits
        loss_fct = nn.CrossEntropyLoss(weight=self.class_weights)

        loss = loss_fct(
            logits.view(-1, model.config.num_labels),
            labels.view(-1)
        )

        return (loss, outputs) if return_outputs else loss

def make_training_args(scale_name):
    TA = TrainingArguments.__init__.__code__.co_varnames
    eval_key = "evaluation_strategy" if "evaluation_strategy" in TA else "eval_strategy"

    kwargs = {
        "output_dir": str(MODELS_DIR / f"distilbert_{scale_name}_weighted"),
        eval_key: "epoch",
        "save_strategy": "epoch",
        "logging_strategy": "steps",
        "logging_steps": 100,
        "num_train_epochs": CONFIG["distilbert_epochs"],
        "per_device_train_batch_size": CONFIG["distilbert_train_batch"],
        "per_device_eval_batch_size": CONFIG["distilbert_eval_batch"],
        "learning_rate": CONFIG["distilbert_lr"],
        "weight_decay": 0.01,
        "seed": SEED,
        "fp16": torch.cuda.is_available(),
        "report_to": "wandb",
        "disable_tqdm": True,
    }

    return TrainingArguments(**kwargs)
```

```python
# Cell 5 — Imports, seeds, and helper functions

import os
import re
import yaml
import time
import json
import math
import random
import hashlib
import joblib
import urllib.request
import numpy as np
import pandas as pd
import torch

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    classification_report,
    precision_recall_curve,
    average_precision_score,
    roc_auc_score,
    f1_score,
    recall_score
)

SEED = CONFIG["seed"]
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

with open(RULES_PATH) as f:
    RULES = yaml.safe_load(f)

def stable_id_from_message(msg: str) -> str:
    if not isinstance(msg, str):
        msg = ""
    return hashlib.sha1(msg[:50000].encode("utf-8", errors="ignore")).hexdigest()

def iter_email_paths(maildir_root):
    for user in os.listdir(maildir_root):
        user_dir = os.path.join(maildir_root, user)
        if not os.path.isdir(user_dir):
            continue
        for root, _, files in os.walk(user_dir):
            for fn in files:
                yield os.path.join(root, fn)

def sample_emails(maildir_root, n=10000, max_chars=20000):
    paths = list(iter_email_paths(maildir_root))
    print("Total email files found:", len(paths))
    chosen = random.sample(paths, n)
    msgs = []
    for p in chosen:
        try:
            with open(p, "r", errors="ignore") as f:
                txt = f.read()
            msgs.append(txt[:max_chars])
        except Exception:
            msgs.append("")
    return pd.DataFrame({"message": msgs})

def extract_subject(text):
    m = re.search(r"(?im)^subject:\s*(.*)$", text or "")
    return m.group(1).strip() if m else ""

def extract_body(text):
    t = text or ""
    parts = t.split("\n\n", 1)
    return parts[1] if len(parts) > 1 else t

def get_header_value(message: str, header: str) -> str:
    if not isinstance(message, str):
        return ""
    m = re.search(rf"(?im)^{header}:\s*(.*)$", message)
    return m.group(1).strip() if m else ""

def extract_addresses(header_value: str):
    if not isinstance(header_value, str):
        return []
    return re.findall(r"[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,}", header_value, flags=re.I)

def has_external_party(message: str) -> bool:
    from_v = get_header_value(message, "From")
    to_v   = get_header_value(message, "To")
    cc_v   = get_header_value(message, "Cc")
    bcc_v  = get_header_value(message, "Bcc")

    addrs = []
    for hv in [from_v, to_v, cc_v, bcc_v]:
        addrs.extend(extract_addresses(hv))

    if not addrs:
        return "@enron.com" not in (message or "").lower()

    return any(not a.lower().endswith("@enron.com") for a in addrs)

def score_phishing(text: str, rules: dict) -> int:
    if not isinstance(text, str) or not text.strip():
        return 0
    t = text.lower()
    weights = rules["weights"]
    score = 0

    core = rules["core_indicators"]
    urgency = rules["urgency_indicators"]
    action = rules["action_indicators"]
    regex_ind = rules["regex_indicators"]
    benign = rules["benign_context_indicators"]

    core_hit = any(k in t for k in core)
    urgency_hit = any(k in t for k in urgency)
    action_hit = any(k in t for k in action)

    url_hit = re.search(regex_ind["url"], t) is not None
    short_hit = re.search(regex_ind["url_shorteners"], t) is not None

    if core_hit:
        score += weights["core"]
    if urgency_hit:
        score += weights["urgency"]
    if action_hit:
        score += weights["action"]
    if url_hit:
        score += weights["url"]
    if short_hit:
        score += weights["url_shortener"]

    benign_hit = any(k in t for k in benign)
    if (not core_hit) and benign_hit and score <= 2:
        return 0
    return score

def is_phish_like_v2(text: str) -> bool:
    threshold = RULES["thresholds"]["phish_score_threshold"]
    return score_phishing(text, RULES) >= threshold

def label_dataset(df):
    df = df.copy()
    df["id"] = df["message"].apply(stable_id_from_message)
    df["subject"] = df["message"].apply(extract_subject)
    df["body"] = df["message"].apply(extract_body)
    df["text"] = df["subject"].fillna("") + " " + df["body"].fillna("")

    def triage(row):
        if is_phish_like_v2(row["text"]):
            return 2
        if has_external_party(row["message"]):
            return 1
        return 0

    df["label3_v3"] = df.apply(triage, axis=1)
    df["y_bin"] = (df["label3_v3"] == 2).astype(int)
    return df

def make_splits_and_save_pt(df, scale_name, out_root=SPLITS_DIR,
                            train_frac=0.80, val_frac=0.10, test_frac=0.10):
    rng = np.random.default_rng(SEED)
    idx = np.arange(len(df))
    rng.shuffle(idx)

    n = len(df)
    n_train = int(n * train_frac)
    n_val   = int(n * val_frac)
    n_test  = n - n_train - n_val

    train_idx = idx[:n_train]
    val_idx   = idx[n_train:n_train+n_val]
    test_idx  = idx[n_train+n_val:]

    keep_cols = ["id", "text", "label3_v3", "y_bin"]

    train_df = df.iloc[train_idx][keep_cols].reset_index(drop=True)
    val_df   = df.iloc[val_idx][keep_cols].reset_index(drop=True)
    test_df  = df.iloc[test_idx][keep_cols].reset_index(drop=True)

    split_dir = Path(out_root) / scale_name
    split_dir.mkdir(parents=True, exist_ok=True)

    train_df[["id"]].to_csv(split_dir / "train_ids.csv", index=False)
    val_df[["id"]].to_csv(split_dir / "val_ids.csv", index=False)
    test_df[["id"]].to_csv(split_dir / "test_ids.csv", index=False)

    torch.save(train_df.to_dict(orient="list"), split_dir / "train.pt")
    torch.save(val_df.to_dict(orient="list"), split_dir / "val.pt")
    torch.save(test_df.to_dict(orient="list"), split_dir / "test.pt")

    meta = {
        "scale": scale_name,
        "seed": SEED,
        "sizes": {"train": len(train_df), "val": len(val_df), "test": len(test_df)},
        "label_distribution_full": df["label3_v3"].value_counts().to_dict(),
        "label_distribution_train": train_df["label3_v3"].value_counts().to_dict(),
        "label_distribution_val": val_df["label3_v3"].value_counts().to_dict(),
        "label_distribution_test": test_df["label3_v3"].value_counts().to_dict(),
    }
    with open(split_dir / "split_meta.yaml", "w") as f:
        yaml.dump(meta, f, sort_keys=False)

    return train_df, val_df, test_df

def load_split_df(scale_name, split):
    d = torch.load(SPLITS_DIR / scale_name / f"{split}.pt")
    return pd.DataFrame(d)
```
